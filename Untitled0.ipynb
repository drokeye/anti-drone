```python
import shutil

def copy_files(src_img_dir, src_lbl_dir, dst_img_dir, dst_lbl_dir, prefix):
    images = glob(f"{src_img_dir}/*.jpg") + glob(f"{src_img_dir}/*.jpeg") + glob(f"{src_img_dir}/*.png")

    for i, img_path in enumerate(images):
        base = os.path.basename(img_path)
        name, ext = os.path.splitext(base)

        new_name = f"{prefix}_{i}{ext}"
        new_lbl  = f"{prefix}_{i}.txt"

        # Copy image
        shutil.copy(img_path, os.path.join(dst_img_dir, new_name))

        # Copy corresponding label
        lbl_path = os.path.join(src_lbl_dir, f"{name}.txt")
        if os.path.exists(lbl_path):
            shutil.copy(lbl_path, os.path.join(dst_lbl_dir, new_lbl))

# Loop through datasets and copy
for dataset, splits in labeled_datasets.items():
    for split, paths in splits.items():
        copy_files(
            paths["img"],
            paths["lbl"],
            f"/content/combined/images/{split}",
            f"/content/combined/labels/{split}",
            prefix=dataset.replace("-", "_")
        )
```

```python
def clean_image_label_folder(img_folder, lbl_folder):
    img_files = os.listdir(img_folder)
    lbl_files = os.listdir(lbl_folder)

    img_stems = {os.path.splitext(f)[0] for f in img_files}
    lbl_stems = {os.path.splitext(f)[0] for f in lbl_files}

    # Remove images with no label
    for fname in img_files:
        stem = os.path.splitext(fname)[0]
        if stem not in lbl_stems:
            os.remove(os.path.join(img_folder, fname))

    # Remove labels with no image
    for fname in lbl_files:
        stem = os.path.splitext(fname)[0]
        if stem not in img_stems:
            os.remove(os.path.join(lbl_folder, fname))

# Apply to both train and val folders
clean_image_label_folder("/content/combined/images/train", "/content/combined/labels/train")
clean_image_label_folder("/content/combined/images/val", "/content/combined/labels/val")
```

```python
import glob
import os

def convert_polygon_to_yolo(txt_file):
    with open(txt_file, "r") as f:
        lines = f.readlines()

    converted_lines = []

    for line in lines:
        parts = line.strip().split()
        if len(parts) < 6 or (len(parts) - 1) % 2 != 0:
            # Skip malformed lines
            continue

        cls = int(parts[0])
        coords = list(map(float, parts[1:]))

        xs = coords[::2]
        ys = coords[1::2]

        x_min = min(xs)
        x_max = max(xs)
        y_min = min(ys)
        y_max = max(ys)

        x_center = (x_min + x_max) / 2
        y_center = (y_min + y_max) / 2
        width = x_max - x_min
        height = y_max - y_min

        converted_line = f"{cls} {x_center:.6f} {y_center:.6f} {width:.6f} {height:.6f}"
        converted_lines.append(converted_line)

    with open(txt_file, "w") as f:
        f.write("\n".join(converted_lines) + "\n")

# 🔁 Convert only BVD files (from bird-vs-drone)
label_files = glob.glob("/content/combined/labels/train/bvd_train_drone_*.txt") + glob.glob("/content/combined/labels/val/bvd_val_drone_*.txt")

print(f"Converting {len(label_files)} polygon label files...")

for txt_file in label_files:
    convert_polygon_to_yolo(txt_file)

print("✅ Conversion complete!")
```

```python
# Adjust if needed
LABELS_PATH = "/content/combined/labels"
IMAGES_PATH = "/content/combined/images"

def draw_yolo_box(img, class_id, x_center, y_center, width, height, color=(0, 255, 0), label="drone"):
    h, w = img.shape[:2]
    x1 = int((x_center - width / 2) * w)
    y1 = int((y_center - height / 2) * h)
    x2 = int((x_center + width / 2) * w)
    y2 = int((y_center + height / 2) * h)

    cv2.rectangle(img, (x1, y1), (x2, y2), color, 2)
    cv2.putText(img, label, (x1, y1 - 5), cv2.FONT_HERSHEY_SIMPLEX, 0.5, color, 1)

def visualize_labelled_image(image_path, label_path):
    image = cv2.imread(image_path)
    if image is None:
        print(f"❌ Failed to load image: {image_path}")
        return

    h, w = image.shape[:2]
    with open(label_path, "r") as f:
        for line in f:
            parts = line.strip().split()
            if len(parts) != 5:
                continue
            cls, xc, yc, ww, hh = map(float, parts)
            draw_yolo_box(image, int(cls), xc, yc, ww, hh)

    image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
    plt.figure(figsize=(8, 8))
    plt.imshow(image)
    plt.axis('off')
    plt.show()

# 🔍 Preview a few from training and validation
from glob import glob
import random

for split in ["train", "val"]:
    label_files = glob(f"{LABELS_PATH}/{split}/bvd_{split}_drone_*.txt")
    if not label_files:
        print(f"No labels found for {split}")
        continue

    print(f"\n🖼️ Visualizing {split.upper()} Set")
    sample = random.choice(label_files)
    img_file = os.path.join(IMAGES_PATH, split, os.path.basename(sample).replace(".txt", ".jpg"))

    visualize_labelled_image(img_file, sample)
```

```python
import os
import shutil
import cv2
from tqdm import tqdm

def move_darknet_dataset_strict(img_list, dst_img_dir, dst_lbl_dir, prefix):
    os.makedirs(dst_img_dir, exist_ok=True)
    os.makedirs(dst_lbl_dir, exist_ok=True)

    copied, skipped = 0, 0

    for i, img_path in tqdm(enumerate(img_list), total=len(img_list), desc=f"Copying {prefix} dataset"):
        name, ext = os.path.splitext(os.path.basename(img_path))
        lbl_path = img_path.replace("Images", "labels").replace(ext, ".txt")

        # === Check image exists, is readable and not empty ===
        if not os.path.exists(img_path) or os.path.getsize(img_path) == 0:
            skipped += 1
            continue

        try:
            img = cv2.imread(img_path)
            if img is None or img.size == 0:
                skipped += 1
                continue
        except:
            skipped += 1
            continue

        # === Check label exists and is not empty ===
        if not os.path.exists(lbl_path) or os.path.getsize(lbl_path) == 0:
            skipped += 1
            continue

        # === Passed all checks, copy files ===
        new_img = f"{prefix}_{i}{ext.lower()}"
        new_lbl = f"{prefix}_{i}.txt"

        shutil.copy2(img_path, os.path.join(dst_img_dir, new_img))
        shutil.copy2(lbl_path, os.path.join(dst_lbl_dir, new_lbl))
        copied += 1

    print(f"\n✅ Copied {copied} valid image-label pairs for prefix '{prefix}'")
    print(f"❌ Skipped {skipped} files (invalid image or empty/missing label)")

def parse_txt_list(txt_path):
    with open(txt_path, 'r') as f:
        lines = [line.strip() for line in f.readlines()]

    return [
        f"/content/datasets/amateur-uav/Database1/Database1/{os.path.basename(line)}"
        for line in lines
    ]

train_imgs = parse_txt_list("/content/datasets/amateur-uav/train.txt")
val_imgs = parse_txt_list("/content/datasets/amateur-uav/test.txt")

move_darknet_dataset_strict(train_imgs, "/content/combined/images/train", "/content/combined/labels/train", prefix="amateur")
move_darknet_dataset_strict(val_imgs, "/content/combined/images/val", "/content/combined/labels/val", prefix="amateur")
```

```python
def clean_yolo_label_file(path):
    cleaned_lines = []
    with open(path, 'r') as f:
        for line in f:
            parts = line.strip().split()
            if len(parts) != 5:
                continue
            cls, x, y, w, h = map(float, parts)
            if not (0 <= x <= 1 and 0 <= y <= 1 and 0 <= w <= 1 and 0 <= h <= 1):
                continue
            cleaned_lines.append(f"{int(cls)} {x:.6f} {y:.6f} {w:.6f} {h:.6f}")

    if cleaned_lines:
        with open(path, 'w') as f:
            f.write('\n'.join(cleaned_lines))
    else:
        os.remove(path)

# Run for all train/val labels
from glob import glob
for path in glob("/content/combined/labels/**/*txt", recursive=True):
    clean_yolo_label_file(path)
```

```python
import os

def remove_orphan_images(img_dir, lbl_dir):
    removed = 0
    for filename in os.listdir(img_dir):
        if not filename.lower().endswith(('.jpg', '.jpeg', '.png')):
            continue

        name, _ = os.path.splitext(filename)
        label_path = os.path.join(lbl_dir, name + ".txt")
        image_path = os.path.join(img_dir, filename)

        if not os.path.exists(label_path) or os.path.getsize(label_path) == 0:
            os.remove(image_path)
            removed += 1

    print(f"✅ Removed {removed} images from {img_dir} with missing/empty labels.")

# Apply to train and val
remove_orphan_images("/content/combined/images/train", "/content/combined/labels/train")
remove_orphan_images("/content/combined/images/val", "/content/combined/labels/val")
```
